# Data Analysis and Visualization with Python:
This notebook demonstrates how to load, analyze, and visualize a dataset using Python libraries like pandas and matplotlib. It includes:
1. Data loading and exploration
2. Basic analysis
3. Visualizations
4. Findings and observations


```python
import pandas as pd

data = pd.read_csv('people_data.csv')

print(data)
```

```python
data['Sex'].value_counts()
```

```python
data.info()
```

```python
data.describe()
```

```python
grouped = data.groupby('Sex').agg({'First Name': 'count'})
print(grouped)
```

```python
import matplotlib.pyplot as plt

data['Year of Birth'] = data['Date of birth'].apply(lambda dob: int(dob.split('-')[0]))

plt.plot(data['Index'], data['Year of Birth'], marker='o', linestyle='-', color='blue')
plt.title('Year of Birth Trends Across Index')
plt.xlabel('Index')
plt.ylabel('Year of Birth')
plt.legend(['Year of Birth'])
plt.grid(True)
plt.show()
```

```python
sex_counts = data['Sex'].value_counts()
sex_counts.plot(kind='bar', color=['pink', 'blue'], edgecolor='black')
plt.title('Count of Individuals by Sex')
plt.xlabel('Sex')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.grid(axis='y')
plt.show()
```

```python
plt.hist(data['Year of Birth'], bins=5, color='green', edgecolor='black')
plt.title('Year of Birth Distribution')
plt.xlabel('Year of Birth')
plt.ylabel('Frequency')
plt.grid(axis='y')
plt.show()
```

```python
plt.scatter(data['Index'], data['Year of Birth'], color='purple')
plt.title('Relationship Between Index and Year of Birth')
plt.xlabel('Index')
plt.ylabel('Year of Birth')
plt.grid(True)
plt.show()
```

```python
import seaborn as sns
sns.histplot(data['Year of Birth'], kde=True, color='red')
plt.title('Year of Birth Distribution (with Density)')
plt.xlabel('Year of Birth')
plt.ylabel('Frequency')
plt.show()
```
